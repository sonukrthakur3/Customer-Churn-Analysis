```python
# 1. Import Required Libraries
import pandas as pd
import numpy as np
```

```python
# 2. Load the Dataset
file_path = "synthetic_customer_churn_100k.csv"
df = pd.read_csv(file_path)

print("✅ Data Loaded Successfully!")
print("-" * 50)
```

```python
# 3. Initial Data Overview
print("📊 Dataset Shape (Rows, Columns):", df.shape)
print("\n📝 First 5 Rows:")
display(df.head())
```

```python
# 4. Data Cleaning Process
print("\n🧹 Starting Data Cleaning...")
```

```python
# a) Convert TotalCharges to numeric (In real-world data, it often contains blank spaces)
# 'coerce' will turn any blank spaces/text into NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
# b) Handle Missing Values
missing_values = df.isnull().sum().sum()
if missing_values > 0:
    print(f"⚠️ Found {missing_values} missing values. Filling with 0 for new customers...")
    # Fill NaN in TotalCharges with 0 (usually happens when Tenure is 0)
    df['TotalCharges'] = df['TotalCharges'].fillna(0)
else:
    print("✨ No missing values found!")
```

```python
# c) Drop Duplicates (if any)
duplicates = df.duplicated().sum()
if duplicates > 0:
    df.drop_duplicates(inplace=True)
    print(f"🗑️ Dropped {duplicates} duplicate rows.")
else:
    print("✨ No duplicate rows found!")
```

```python
# d) Standardize Text Columns (Best practice for Power BI)
# Making categorical data uniform
df['Gender'] = df['Gender'].str.capitalize()
df['Contract'] = df['Contract'].str.title()
df['PaymentMethod'] = df['PaymentMethod'].str.title()

print("\n✅ Data Cleaning Completed!")
print("-" * 50)
```

```python
# 5. Quick Exploratory Data Analysis (EDA)
print("📈 Quick Churn Analysis:")
churn_counts = df['Churn'].value_counts()
churn_rate = (churn_counts['Yes'] / len(df)) * 100

print(f"Total Customers: {len(df)}")
print(f"Churned Customers: {churn_counts.get('Yes', 0)}")
print(f"Overall Churn Rate: {churn_rate:.2f}%")
```

```python
# 6. Export Cleaned Data for Power BI
cleaned_file_name = "Cleaned_Churn_Data.csv"
df.to_csv(cleaned_file_name, index=False)

print("-" * 50)
print(f"🎉 Success! Cleaned dataset saved as '{cleaned_file_name}'.")
print("📥 You can now download this file from Colab and load it into Power BI.")
```
